```python
<VSCode.Cell id="1" language="markdown">
# Gradient Descent Simulation Exercise

This notebook provides a framework for students to simulate and compare gradient descent methods that use only function evaluations versus those using function and gradient evaluations. The exercise aims to analyze efficiency and accuracy.
</VSCode.Cell>

<VSCode.Cell id="2" language="markdown">
## 1. Create a Python Module for the Exercise Class

Define classes and functions to encapsulate the simulation of gradient descent, including counting evaluation calls.
</VSCode.Cell>

<VSCode.Cell id="3" language="python">
class EvaluationCounter:
    def __init__(self):
        self.count_func = 0
        self.count_grad = 0

    def reset(self):
        self.count_func = 0
        self.count_grad = 0

    def eval_func(self, f, x):
        self.count_func += 1
        return f(x)

    def eval_grad(self, grad_f, x):
        self.count_grad += 1
        return grad_f(x)


# Example class to encapsulate functions and their evaluations
class FunctionWithCounter:
    def __init__(self, func, grad=None):
        self.func = func
        self.grad = grad
        self.counter = EvaluationCounter()
    
    def evaluate(self, x):
        return self.counter.eval_func(self.func, x)
    
    def evaluate_grad(self, x):
        if self.grad:
            return self.counter.eval_grad(self.grad, x)
        else:
            raise NotImplementedError("Gradient not provided")

</VSCode.Cell>

<VSCode.Cell id="4" language="markdown">
## 2. Numerical Gradient Approximation

Use finite differences to estimate derivatives, involving two function calls per variable, and count evaluations.
</VSCode.Cell>

<VSCode.Cell id="5" language="python">
def numerical_gradient(f_obj, x, epsilon=1e-6):
    grad =
    for i in range(len(x)):
        x1 = x.copy()
        x2 = x.copy()
        x1[i] += epsilon
        x2[i] -= epsilon
        f_x1 = f_obj.evaluate(x1)
        f_x2 = f_obj.evaluate(x2)
        grad_element = (f_x1 - f_x2) / (2 * epsilon)
        if 'grad' not in locals():
            grad = [grad_element]
        else:
            grad.append(grad_element)
    return grad

</VSCode.Cell>

<VSCode.Cell id="6" language="markdown">
## 3. Gradient Descent Using Analytic Gradient

Leverage provided gradients to reduce the number of function evaluations during descent.
</VSCode.Cell>

<VSCode.Cell id="7" language="python">
def gradient_descent_with_grad(f_obj, grad_f, x0, learning_rate=0.01, max_iter=1000, tol=1e-6):
    x = x0.copy()
    evaluations = 0
    for i in range(max_iter):
        grad = f_obj.evaluate_grad(x)
        x_new = [x_i - learning_rate * g for x_i, g in zip(x, grad)]
        if abs(f_obj.evaluate(x_new) - f_obj.evaluate(x)) < tol:
            break
        x = x_new
        evaluations += 1
    return x, evaluations

</VSCode.Cell>

<VSCode.Cell id="8" language="markdown">
## 4. Count Evaluation Calls

Maintain counters during simulations and report total calls for comparison.
</VSCode.Cell>

<VSCode.Cell id="9" language="python">
def run_simulation_example():
    # Example function
    def func(x):
        return (x[0] - 1)**2 + (x[1] - 2)**2
    
    def grad(x):
        return [2 * (x[0] - 1), 2 * (x[1] - 2)]

    f_obj = FunctionWithCounter(func, grad)
    x0 = [0, 0]

    # Simulate using finite differences
    f_obj.counter.reset()
    grad_fd = numerical_gradient(f_obj, x0)
    evals_fd = f_obj.counter.count_func

    # Simulate using analytic gradient
    f_obj.counter.reset()
    final_x, evals_grad = gradient_descent_with_grad(f_obj, grad, x0)

    print(f"Function calls (finite differences): {evals_fd}")
    print(f"Function calls (analytic gradient): {evals_grad}")
    print(f"Final position: {final_x}")

run_simulation_example()

</VSCode.Cell>

<VSCode.Cell id="10" language="markdown">
## 5. Reporting and Comparison

Compare convergence speed, total evaluations, and accuracy.
</VSCode.Cell>

<VSCode.Cell id="11" language="python">
def report_comparison():
    # This function will compare the simulation results
    pass  # Implementation details to be added

</VSCode.Cell>

<VSCode.Cell id="12" language="markdown">
## 6. Example Usage

Demonstrate simulations with sample functions and report on performance.
</VSCode.Cell>

<VSCode.Cell id="13" language="python">
# Example usage
# Define your test functions and run simulation reports.

</VSCode.Cell>
```
